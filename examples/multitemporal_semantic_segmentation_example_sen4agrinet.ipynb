# Example of the aitlas toolbox in the context of semantic segmentation

The Sen4AgriNet dataset is multi-temporal. This notebook shows how to load the images, visualize them with their labels, obtain statistics and plot a time-series graph.

## Import the required packages

```python
from aitlas.datasets import Sen4AgriNetDataset
from aitlas.utils import image_loader
from aitlas.models import DeepLabV3
from aitlas.transforms import MinMaxNormTranspose
```

## Visualize images and masks

```python
dataset_config = {
    "data_dir": "./data/sen4agrinet/data",
    "selection":"rgb", #rgb or all 13 bands
    "date": "1", #visualise images for a particular date
    "class_selection": "12" #select all 168 classes or the 12 classes used in the experiments in doi.org/10.1109/JSTARS.2022.3164771
}
dataset = Sen4AgriNetDataset(dataset_config) 

print(f"Total number of patches: {len(dataset)}")
```

```python
fig1 = dataset.show_image(0)
fig2 = dataset.show_image(300)
fig3 = dataset.show_image(500)
```

```python
dataset.data_distribution_table()
```

```python
fig = dataset.data_distribution_barchart()
```

```python
dataset_config = {
    "data_dir": "./data/sen4agrinet/data",
    "selection":"all",
    "class_selection": "12" 
}
dataset = Sen4AgriNetDataset(dataset_config)

fig = dataset.show_timeseries(300, 150, 150)
```
