# Example of the aitlas toolbox in the context of semantic segmentation

The CropHarvest dataset is multi-temporal. The labels are stored in a GeoJSON file. This notebook shows how to load the images, visualize them with their labels, obtain statistics and plot a time-series graph.

## Import the required packages

```python
from aitlas.datasets import CropHarvestDataset
from aitlas.models import LSTM
from aitlas.transforms import ResizeCenterCropFlipHVToTensor, ResizeCenterCropToTensor, ResizeRandomCropFlipHVToTensor
from aitlas.utils import image_loader
```

## Load the dataset

```python
dataset_config = {
    "data_dir": "./data/cropharvest",
}
dataset = CropHarvestDataset(dataset_config)

print(f"Total number of patches: {len(dataset)}")
```

## Show images from the dataset

```python
fig1 = dataset.show_image(100)
fig2 = dataset.show_image(2000)
fig3 = dataset.show_image(8000);
```

## Inspect the data

```python
dataset.data_distribution_table()
```

```python
fig = dataset.data_distribution_barchart();
```

## Time series plot

```python
fig = dataset.show_timeseries(index=400, longitude=10, latitude=10)
```
