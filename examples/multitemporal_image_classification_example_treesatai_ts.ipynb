# Example of the aitlas toolbox in the context of multi class image classification

The TreeSatAI_TS dataset is multi-temporal. This notebook shows how to load the images, visualize them with their labels, obtain statistics and plot a time-series graph.

## Import the required packages

```python
import sys
sys.path.insert(0, "..")  # makes the datasets package of this repository importable
from datasets import TreeSatAI_TSDataset
from aitlas.models import LSTM
from aitlas.transforms import ResizeCenterCropFlipHVToTensor, ResizeCenterCropToTensor, ResizeRandomCropFlipHVToTensor
from aitlas.utils import image_loader
```

## Load the dataset

```python
dataset_config = {
    "data_dir": "./data/TreeSatAI-ts",
    "csv_file": "./data/TreeSatAI-ts/train_filenames.lst",
}
dataset = TreeSatAI_TSDataset(dataset_config)

print(f"Total number of patches: {len(dataset)}")
```
