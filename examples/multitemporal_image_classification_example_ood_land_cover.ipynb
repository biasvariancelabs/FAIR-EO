# Example of the aitlas toolbox in the context of image classification

The Out-of-Domain Land Cover dataset is a multi-temporal dataset. It contains a single value per MODIS band for each time step.

```python
from aitlas.datasets import OODLandCoverDataset
from aitlas.models import LSTM 
from aitlas.utils import image_loader
```
