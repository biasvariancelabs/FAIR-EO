# Example of the aitlas toolbox in the context of semantic segmentation

The MTS12 dataset is multi-temporal. This notebook shows how to load the images, visualize them with their labels, obtain statistics and plot a time-series graph.

## Import the required packages

```python
from aitlas.datasets import MTS12Dataset
from aitlas.utils import image_loader
from aitlas.models import DeepLabV3
from aitlas.transforms import MinMaxNormTranspose
```

## Visualize images and masks

### Sentinel-2

```python
dataset_config = {
    "data_dir": "./data/mts12/",
    "selection":"rgb", #rgb or all Sentinel-2 bands available
    "date": "6", #visualise images for a particular date
    "imagery" : "s2", #select which Sentinel imagery to use
    "split": "train"
}
dataset = MTS12Dataset(dataset_config) 

print(f"Total number of patches: {len(dataset)}")
```

```python
fig1 = dataset.show_image(20)
fig2 = dataset.show_image(290)
fig3 = dataset.show_image(320);
```

```python
dataset.data_distribution_table()
```

```python
fig = dataset.data_distribution_barchart();
```

```python
dataset_config = {
    "data_dir": "./data/mts12/",
    "selection": "all", 
    "date": None, #None for timeseries
    "imagery" : "s2", 
    "split": "train"
}
dataset = MTS12Dataset(dataset_config) 

fig = dataset.show_timeseries(300, 100, 50)
```
