# Example of the aitlas toolbox in the context of semantic segmentation

The CV4A Kenya Crop Type dataset is multi-temporal. This notebook shows how to load the images, visualize them with their labels, obtain statistics and plot a time-series graph.

## Import the required packages

```python
import sys
sys.path.insert(0, "..")  # makes the datasets package of this repository importable
from datasets import CV4AKenyaCropTypeDataset
from aitlas.models import LSTM
from aitlas.transforms import ResizeCenterCropFlipHVToTensor, ResizeCenterCropToTensor, ResizeRandomCropFlipHVToTensor
from aitlas.utils import image_loader
```

## Load the dataset

```python
dataset_config = {
    "data_dir": "./data/cv4a_kenya_crop_type",
}
dataset = CV4AKenyaCropTypeDataset(dataset_config)

print(f"Total number of images: {len(dataset)}. Total number of labels: {len(dataset.annotations)}.")
```

## Show images from the dataset

```python
fig1 = dataset.show_image(2)
fig2 = dataset.show_image(3);
```

## Inspect the data

```python
dataset.data_distribution_table()
```

```python
fig = dataset.data_distribution_barchart(); # the no-data value is ignored
```

## Time series plot

```python
fig = dataset.show_timeseries(index=2, longitude=1, latitude=1)
```
