# Example of the aitlas toolbox in the context of image segmentation

This notebook shows a sample implementation of semantic segmentation using the aitlas toolbox. The Crop Type Mapping (Ghana and South Sudan) dataset contains time series images of Sentinel-1, Sentinel-2, PlanetScope. Along with the corresponding label mask.

## Import the required packages

```python
from aitlas.datasets import CropTypeGhanaSoudanDataset
from aitlas.utils import image_loader
from aitlas.models import DeepLabV3
from aitlas.transforms import MinMaxNormTranspose, ResizeToTensor
```

## Visualize images and masks

```python
dataset_config = {
    "data_dir" : "./data/crop_type_mapping_ghana_south_sudan",
}
dataset = CropTypeGhanaSoudanDataset(dataset_config)

print(f"Total number of patches: {len(dataset)}")
```

# Show images
